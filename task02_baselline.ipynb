```python
import pandas as pd

df = pd.read_csv("customer-churn-training.csv")

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nTarget Distribution:")
print(df["churned"].value_counts())
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Majority-class baseline
majority_class = df["churned"].mode()[0]

df["baseline_prediction"] = majority_class

accuracy = accuracy_score(df["churned"], df["baseline_prediction"])

print("Majority Class:", majority_class)
print("Baseline Accuracy:", round(accuracy, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(df["churned"], df["baseline_prediction"]))

print("\nClassification Report:")
print(classification_report(
    df["churned"],
    df["baseline_prediction"],
    zero_division=0
))
```

```python
print("Dataset Summary")
print("=" * 40)

print(f"Number of customers: {len(df)}")
print(f"Number of features: {len(df.columns) - 2}")  # excluding ID and target
print(f"Churned customers: {(df['churned'] == 1).sum()}")
print(f"Non-churned customers: {(df['churned'] == 0).sum()}")

print("\nFeature Data Types:")
print(df.dtypes)

print("\nUnique Plan Types:")
print(df["plan_type"].unique())
```

```python
print("Feature Ranges")
print("=" * 40)

for col in [
    "tenure_months",
    "support_tickets",
    "monthly_spend_inr",
    "last_login_days"
]:
    print(f"{col}: {df[col].min()} to {df[col].max()}")

print("\nChurn Rate:", round(df["churned"].mean() * 100, 2), "%")
```
